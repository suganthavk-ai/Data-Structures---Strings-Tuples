# **1. String Concatenation**

```python
# Get the name from the user
name = input("Enter Your Name: ")

# Concatenate "Hello" with the name
str1 = "Hello " + name
print(str1)

# Add the welcome message
str3 = ", welcome to Python programming"
result = str1 + str3

print(result)
```

# **2. String Slicing and Indexing**
Using the concatenated string:

a. Print the first character of the string.

b. Print the last character of the string.

c. Print the first 5 characters of the string.

d. Print the last 11 characters of the string.

e. Print the string in reverse.

f. Use slicing and print the word “Python” from the existing string.


```python
name = input("Enter Your Name: ")

str1 = "Hello " + name
result = str1 + ", Welcome to Python Programming"

print("Oringinal String: ", result)
```

```python
# a. Print the first character of the string.
print("First Character: ", result[0])
```

```python
# b. Print the last character of the string.
print("Last Character: ", result[-1])
```

```python
# c. Print the first 5 characters of the string.
print("First 5 Character: ", result[:5])
```

```python
# d. Print the last 11 characters of the string.
print("Last 11 characters:", result[-11:])
```

```python
# e. Print the string in reverse.
print("Reverse: ",result[::-1])
```

```python
# f. Use slicing and print the word “Python” from the existing string.
print("Python:", result[ result.index("Python") : result.index("Python") + 6])
```

# **3. String Methods:**

Write a Python program that takes a string, strM = “Python beginner tutorial," and performs the following tasks:

a. Convert the sentence to uppercase.

b. Convert the sentence to lowercase.

c. Use Capitalize and return the sentence to the original input form.

d. Count the total number of occurrences of character ‘t’ in the string.

e. Replace all occurrences of “Python” with “Data Analytics” in the input string strM = “Python beginner tutorial.”


```python
strM = "Python beginner tutorial"
# a. Convert the sentence to uppercase.
print("UpperCase: ",strM.upper())
```

```python
# b. Convert the sentence to lowercase.
print("LowerCase: ",strM.lower())
```

```python
# c. Use Capitalize and return the sentence to the original input form.
print("Capitalize: ", strM.capitalize())
```

```python
# d. Count the total number of occurrences of character ‘t’ in the string.
print("Count of 't' : ", strM.count("t"))
```

```python
# e. Replace all occurrences of “Python” with “Data Analytics” in the input string strM = “Python beginner tutorial.”
strA = strM.replace("Python", "Data Analytics")
print("New String : ", strA)
```

# **Tuples (Creation, Modification and Access) :**

Create
the 1st tuple with values -> (10, 20, 30) and
the 2nd tuple with values -> (40, 50, 60):

a. Concatenate the two tuples and store it in “t_combine”

b. Repeat the elements of “t_combine” 3 times

c. Access the 3rd element from “t_combine”

d. Access the first three elements from “t_combine”

e. Access the last three elements from “t_combine”


```python
t1 = (10, 20, 30)
t2 = (40, 50, 60)

# a. Concatenate the two tuples and store it in “t_combine”
t_combine = t1 + t2
print("Combined Tuple: ", t_combine)
```

```python
# b. Repeat the elements of “t_combine” 3 times
t_repeat = t_combine * 3
print("Repeated Tuple: ", t_repeat)
```

```python
# c. Access the 3rd element from “t_combine”
print("3rd element: ", t_combine[2])
```

```python
# d. Access the first three elements from “t_combine”
print("First Three Element: ", t_combine[:3])
```

```python
# e. Access the last three elements from “t_combine”
print("Last Three Element: ", t_combine[-3:])
```
